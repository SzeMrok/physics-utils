```python
from physics_utils.orbital import (
    escape_velocity,
    orbital_velocity,
    orbital_period
)

from physics_utils.mechanics import kinetic_energy
```

```python
# Earth escape velocity

earth_mass = 5.972e24
earth_radius = 6.371e6

v = escape_velocity(earth_mass, earth_radius)

print(v.to("km/s"))
```

```python
# Orbital velocity at 400 km altitude

v_orbit = orbital_velocity(earth_mass, earth_radius + 400e3)

print(v_orbit.to("km/s"))
```

```python
# Orbital period at 400 km altitude

T = orbital_period(earth_mass, earth_radius + 400e3)

print(T.to("minute"))
```

```python
# kinetic energy of a 1000 kg satellite in low Earth orbit

e = kinetic_energy(1000, v_orbit)

print(e.to("MJ"))
```
